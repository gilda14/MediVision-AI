```python

message = "Hello, I want to change my booking date."

message = message.lower()

if "broken" in message or "breakdown" in message:
    category = "Vehicle Breakdown"
    priority = "HIGH"

elif "price" in message or "cost" in message:
    category = "Rental Price Enquiry"
    priority = "NORMAL"

elif "booking" in message or "change" in message:
    category = "Booking Change"
    priority = "MEDIUM"

else:
    category = "Other Enquiry"
    priority = "NORMAL"

print("Category:", category)
print("Priority:", priority)
```

```python

# MediVision AI - Lesson 1
# Fictional patient data for learning purposes

medical_report = {
    "patient_id": "DEMO-001",
    "report_date": "2026-11-11",
    "test_results": {
        "Haemoglobin": {
            "value": 142,
            "unit": "g/L"
        },
        "Glucose": {
            "value": 5.8,
            "unit": "mmol/L"
        },
        "Cholesterol": {
            "value": 4.2,
            "unit": "mmol/L"
        }
    }
}

print("Patient:", medical_report["patient_id"])
print("Date:", medical_report["report_date"])

for test, result in medical_report["test_results"].items():
    print(test, result["value"], result["unit"])
```

Create a reusable report function

```python

def create_medical_summary(report):
    lines = []

    lines.append("MEDIVISION AI REPORT")
    lines.append("Patient: " + report["patient_id"])
    lines.append("Date: " + report["report_date"])
    lines.append("")

    for test, result in report["test_results"].items():
        line = (
            f"{test}: {result['value']} "
            f"{result['unit']}"
        )
        lines.append(line)

    return "\n".join(lines)


summary = create_medical_summary(medical_report)

print(summary)
with open("medical_summary.txt", "w") as file:
    file.write(summary)

print("Medical summary saved successfully!")
```
