```python

patient_name = "Test Patient"
patient_age = 45

print("Patient:", patient_name)
print("Age:", patient_age)
```

```python

# MediVision AI - Patient Medical Record

patient = {
    "name": "Test Patient",
    "age": 45,
    "blood_pressure": "120/80",
    "heart_rate": 72,
    "temperature": 36.8
}

print(patient)
```

```python

# MediVision AI - Retrieve Patient Information

print("Patient Name:", patient["name"])

print("Blood Pressure:", patient["blood_pressure"])

print("Heart Rate:", patient["heart_rate"])
```

```python

# MediVision AI - Add Laboratory Results

patient["glucose"] = 5.8

patient["cholesterol"] = 4.2

print("Patient:", patient["name"])
print("Glucose:", patient["glucose"], "mmol/L")
print("Cholesterol:", patient["cholesterol"], "mmol/L")
```

This Python program can store patient information and laboratory results.
